```python
from Inventory.Inventory_models import InventoryParams
from Inventory.inventory_simulation import InventorySimulation
```

```python
params = InventoryParams(
    D=3650,
    T_total=365,
    LD=2,
    T=5,
    Q=50,
    initial_ioh=50,
    sigma=2.5,
    S=500,
    H=20
)

simulation = InventorySimulation(params)
```

```python
result_1 = simulation.simulation_1()

result_1.head(10)
```

```python
result_2 = simulation.simulation_2()

result_2.head(10)
```

```python
result_3 = simulation.simulation_3()

result_3.head(10)
```

```python
result_4 = simulation.simulation_4()

result_4.head(10)
```

```python
result_5 = simulation.simulation_5()

result_5.head(10)
```

```python
result_6 = simulation.simulation_6()

result_6.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 7))

plt.plot(
    result_1["time"],
    result_1["inventory_level"],
    label="Simulation 1"
)

plt.plot(
    result_2["time"],
    result_2["inventory_level"],
    label="Simulation 2"
)

plt.plot(
    result_3["time"],
    result_3["inventory_level"],
    label="Simulation 3"
)

plt.plot(
    result_4["time"],
    result_4["inventory_level"],
    label="Simulation 4"
)

plt.plot(
    result_5["time"],
    result_5["inventory_level"],
    label="Simulation 5"
)

plt.plot(
    result_6["time"],
    result_6["inventory_level"],
    label="Simulation 6"
)

plt.axhline(
    y=0,
    linestyle="--",
    label="Zero Inventory"
)

plt.xlabel("Time (Days)")
plt.ylabel("Inventory Level")

plt.title("Comparison of Inventory Simulations")

plt.legend()

plt.grid(True)

plt.show()
```
